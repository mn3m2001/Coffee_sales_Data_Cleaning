```python
import pandas as pd 

file_path = r"M:\Git Hub\Git Hub Project\coffee Sales dataset\Coffe_sales.csv"
Coffee_data = pd.read_csv(file_path)
print(Coffee_data.head()
)
```

```python
#number of rows and columns
print (Coffee_data.shape)
Coffee_data.info()
```

```python
print (Coffee_data.isna().sum()) #to check for missing values in the dataset
```

```python
print(Coffee_data.duplicated().sum()) #to check for duplicate values in the dataset
```

```python
#make a copy of data to compare with the original data after cleaning
Coffee_data_cleaned = Coffee_data.copy()
# اعرض اسماء الاعمده 
print(Coffee_data_cleaned.columns.tolist())
# تغير شكل الاسماء الاعمده لتسهيل التعامل معها
Coffee_data_cleaned.columns = Coffee_data_cleaned.columns.str.strip().str.lower()

```

```python
print (Coffee_data_cleaned["coffee_name"].value_counts()) #to check for unique values in the coffee_name column
```

```python
#to convert the date column to datetime format, رتب القيم علي حسب التاريخ وغير اسماء التواريخ الي ارقام فعليه 
parsed_date = pd.to_datetime(
    Coffee_data_cleaned['date'],
    errors='coerce',
    format='%Y-%m-%d'
    ) 
print (parsed_date)
```

```python
invalid_dates = Coffee_data_cleaned.loc[
    parsed_date.isna() & Coffee_data_cleaned["date"].notna(),
    ["date"]
]
print(invalid_dates)
```

```python
#بعد ما بتأكد ان ميش تواريخ ناقصه بنقل القيم الجديده للتواريخ للجدول الاصلي عندي 

Coffee_data_cleaned['date'] = parsed_date
#اتاكد من نوع العمود لان المفروض انا حولته من str to int    
print (Coffee_data_cleaned['date'].dtype)
```

```python
#هتأكد انت التواريخ مطابقه مع ال month short 
month_from_date = Coffee_data_cleaned['date'].dt.month
print(month_from_date)
```

```python
# هقارن بين عمود التواريخ الاصلي وعمود التاريخ المعدل هل فيه اختلاف ولا لا 


month_mismatch = Coffee_data_cleaned["monthsort"] != month_from_date

print(Coffee_data_cleaned.loc[month_mismatch, ["date", "monthsort"]])
```

```python
#هشوف الصفوف الي رقم الشهر فيها مختلف عن التاريخ 
month_errors = Coffee_data_cleaned.loc[month_mismatch, ["date", "monthsort"]]
print(month_errors)
```
